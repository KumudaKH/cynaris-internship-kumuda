## Exploratory Data Analysis using a Real Dataset

```python
import pandas as pd
```

```python
df = pd.read_csv('Downloads/archive/tech_mental_health_burnout.csv')
```

```python
print("Dataset loaded successfully")
df.head()
```

```python
df.shape
```

```python
print("\nDataset information:")
df.info()
```

```python
print("\nStatistical summary:")
df.describe()
```

```python
# 3. Count categorical values

print("\nPassenger count by gender:")
print(df["gender"].value_counts())
```

```python
print("\nEmployees by Job Role:")
df["job_role"].value_counts()
```

```python
print("\nBurnout Level Distribution:")
df["burnout_level"].value_counts()
```

```python
# 4. Select columns
print("\nSelected Columns:")
df[["age", "gender", "job_role", "stress_level", "burnout_level"]].head()
```

```python
# Filter employees with high burnout scores
print("\nEmployees with Burnout Score Above 2:")
df[df["burnout_score"] > 2.0].head()
```

```python
# Filter employees working more than 40 hours per week
print("\nEmployees Working More Than 40 Hours:")
df[df["work_hours_per_week"] > 40].head()
```

```python
# 5. Handle missing values
print("\nMissing Values Before Cleaning:")
df.isnull().sum()
```

```python
# Fill missing numerical values with their column median
for col in df.select_dtypes(include="number").columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
# Fill missing categorical values with their mode
for col in df.select_dtypes(include="object").columns:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

```python
# 6. Group by job role and aggregate
print("\nAverage Stress and Burnout by Job Role:")
df.groupby("job_role").agg(
        Employee_Count=("age", "count"),
        Average_Stress=("stress_level", "mean"),
        Average_Burnout=("burnout_score", "mean"),
        Average_Sleep=("sleep_hours", "mean")
    ).round(2)
```

```python
# Group by work mode
print("\nAverage Burnout by Work Mode:")
print(
    df.groupby("work_mode")["burnout_score"]
    .mean()
    .round(2)
)
```

```python
# 7. Demonstrate loc and iloc
print("\nUsing loc:")
print(df.loc[0, ["age", "gender", "job_role"]])

print("\nUsing iloc:")
df.iloc[0:5, 0:5]
```

```python
# 8. Export cleaned data
df.to_csv("cleaned_employee_burnout.csv", index=False)

print("\nCleaned dataset exported successfully!")
```
