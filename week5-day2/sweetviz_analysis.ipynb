```python
import pandas as pd
import sweetviz as sv
import os

print("SweetViz imported successfully!")
```

```python
# Load the W5D2 Salary Dataset

# Load dataset
df = pd.read_csv(r"C:\Users\KUMUDHA\OneDrive\Desktop\cynaris-internship-kumuda\week5-day2\Salary_Dataset_with_Extra_Features.csv")

# Display first five rows
display(df.head())
print("Dataset Shape:", df.shape)

display(df.head())
```

```python
# Dataset overview

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
display(df.describe(include="all"))
```

```python
# Generate SweetViz automated EDA report

report = sv.analyze(df)

# Save HTML report inside reports folder
report.show_html(
    filepath="reports/SweetViz_Report.html",
    open_browser=False
)

print("SweetViz report generated successfully!")
```

```python
# Verify generated HTML report

report_path = "reports/SweetViz_Report.html"

if os.path.exists(report_path):
    print("Report successfully generated!")
    print("File Size:", round(os.path.getsize(report_path) / 1024, 2), "KB")
else:
    print("Report generation failed!")
```

```python
import sweetviz as sv
import pandas as pd
import os

df = pd.read_csv(r"C:\Users\KUMUDHA\OneDrive\Desktop\cynaris-internship-kumuda\week5-day2\Salary_Dataset_with_Extra_Features.csv")

report = sv.analyze(df)

output_path = os.path.abspath("reports/SweetViz_Report.html")

os.makedirs(os.path.dirname(output_path), exist_ok=True)

report.show_html(
    filepath=output_path,
    open_browser=False
)

print("Report generated:", os.path.exists(output_path))
print("Saved at:", output_path)
```
