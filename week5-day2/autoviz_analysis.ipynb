```python
import os

# Create output directory
report_dir = "reports/autoviz"
os.makedirs(report_dir, exist_ok=True)

# Run AutoViz
dft = AV.AutoViz(
    filename="",
    sep=",",
    depVar="",
    dfte=df,
    header=0,
    verbose=1,
    lowess=False,
    chart_format="html",
    max_rows_analyzed=22770,
    max_cols_analyzed=8,
    save_plot_dir=report_dir
)

print("AutoViz analysis completed!")
print("Reports saved in:", report_dir)
```

```python
# Salary distribution analysis

print("SALARY DISTRIBUTION")

print("Average Salary:", df["Salary"].mean())
print("Median Salary:", df["Salary"].median())
print("Minimum Salary:", df["Salary"].min())
print("Maximum Salary:", df["Salary"].max())

print("\nSalary Skewness:", df["Salary"].skew())
```

```python
# Average salary by job role

job_salary = df.groupby("Job Roles")["Salary"].agg(
    ["mean", "median", "count"]
).sort_values(by="mean", ascending=False)

display(job_salary.head(10))
```

```python
# Average salary by location

location_salary = df.groupby("Location")["Salary"].agg(
    ["mean", "median", "count"]
).sort_values(by="mean", ascending=False)

display(location_salary)
```

```python
# Employment status distribution

employment_analysis = df.groupby("Employment Status").agg(
    Employee_Count=("Salary", "count"),
    Average_Salary=("Salary", "mean")
).sort_values(by="Employee_Count", ascending=False)

display(employment_analysis)
```

Mumbai records the highest average salary of approximately ₹9.61 lakh, followed by Bangalore at ₹7.35 lakh. Bangalore contributes the highest number of salary records, with 8,264 observations.
