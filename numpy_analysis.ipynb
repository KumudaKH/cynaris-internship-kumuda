```python
salaries = np.array([
    25000, 35000, 45000, 55000,
    65000, 75000, 85000, 95000
])

# Boolean masking
high_salaries = salaries[salaries > 60000]

print("Salaries above 60000:", high_salaries)

# Fancy indexing
selected_salaries = salaries[[0, 3, 5, 7]]

print("Selected salaries:", selected_salaries)

# Multiple conditions
medium_salaries = salaries[
    (salaries >= 40000) & (salaries <= 80000)
]

print("Salaries between 40000 and 80000:", medium_salaries)
```

```python
# Create one million numbers
n = 1_000_000

numbers_list = list(range(n))
numbers_array = np.arange(n)

# Python loop benchmark
start = time.perf_counter()

python_result = [x * x for x in numbers_list]

python_time = time.perf_counter() - start

# NumPy vectorised benchmark
start = time.perf_counter()

numpy_result = numbers_array * numbers_array

numpy_time = time.perf_counter() - start

# Results
print(f"Python loop time: {python_time:.6f} seconds")
print(f"NumPy vectorised time: {numpy_time:.6f} seconds")

if numpy_time > 0:
    print(
        f"NumPy speedup: {python_time / numpy_time:.2f}x"
    )

# Verify that both approaches produce the same result
print(
    "Results match:",
    np.array_equal(
        np.asarray(python_result),
        numpy_result
    )
)
```

```python
benchmark_results = pd.DataFrame({
    "Method": ["Python List Comprehension", "NumPy Vectorisation"],
    "Execution Time (seconds)": [python_time, numpy_time]
})

display(benchmark_results)

benchmark_results.to_csv(
    "numpy_benchmark_results.csv",
    index=False
)

print("Benchmark results saved.")
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    benchmark_results["Method"],
    benchmark_results["Execution Time (seconds)"]
)

plt.title("Python Loop vs NumPy Performance")
plt.ylabel("Execution Time (seconds)")
plt.xlabel("Method")
plt.xticks(rotation=10)
plt.tight_layout()
plt.show()
```
